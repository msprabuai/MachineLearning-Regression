```python
# Salary Prediction Phase1 using SLR Model Creation.
#Pandas Pandas is a Python library used to store, organize, clean, analyze, and manipulate data easily, similar to working with Excel spreadsheets.
import pandas as pd #(pandas = pd)
```

```python
dataset=pd.read_csv("Salary_Data.csv") #Calling source csv file and should place the csv file in the same path where you have this place.
```

```python
dataset #Execute the csv file to view the dataset.
```

```python
independent=dataset[["years_of_experience"]] #Invoke the input column values and assiging to the variable named as independent x  input
independent
```

```python
dependent= dataset ["monthly_salary_inr"] #Invoke the input column values and assiging to the variable named as dependent y output
dependent
```

```python
#Split Training Set and Test Set.
from sklearn.model_selection import train_test_split #Scikit-learn, a Python Machine Learning library, ,model_selection - A module used to select and evaluate ML models, train_test_split - A function used to split data into training and testing datasets  
x_train,x_test,y_train,y_test=train_test_split(independent,dependent,test_size=0.30, random_state=0) #x_train (input training),x_test (input testing),y_train(output training), y_test (output testing)
```

```python
from sklearn.linear_model import LinearRegression #used to import the Linear Regression algorithm from the Scikit-le0arn library
regressor = LinearRegression() # model creation - linearregression considered as a class assigned to the variable regressor.
regressor.fit(x_train,y_train) #Train the Linear Regression model using the training input data and its corresponding actual output values.
```

```python
# Extract the weight (slope/coefficient) learned by the Linear Regression model.
# It represents how much the predicted output (Salary) changes when the input
# feature (Experience) increases by 1 unit.
weight=regressor.coef_
weight
```

```python
# It is the predicted Salary when Experience is 0
bais=regressor.intercept_
bais
```

```python
# Use the trained model to predict Salary for the test input data
# x_test contains the Experience values that the model has not used for training
y_pred=regressor.predict(x_test)
```

```python
# Import the R² score evaluation function from sklearn
from sklearn.metrics import r2_score

# Compare actual Salary values (y_test) with predicted Salary values (y_pred)
# R² indicates how well the model explains the variation in the actual values
r2_score=r2_score(y_test,y_pred)

# Display the R² score
r2_score
```

```python
# Import the pickle library.
# Pickle is used to save Python objects into a file and load them later.
import pickle 
# Specify the name of the file where the trained model will be saved.
# The .sav extension is commonly used for saved model files.
filename="finalized_model_linear.sav"
```

```python
# Save the trained Linear Regression model into the specified file.
# regressor  = the trained model
# open()     = opens the file
# filename   = name of the file
# 'wb'       = write in binary mode
# pickle.dump() = converts and writes the model object into the file
pickle.dump(regressor,open(filename,'wb'))
```

```python
# Validate whether the saved model can be loaded and used for prediction.
# If successful, we can proceed to the next phase of deployment.
# Open the saved model file in read-binary ('rb') mode and load it.
# pickle.load() reads the saved model and restores it as a Python object.
loaded_model = pickle.load(open("finalized_model_linear.sav",'rb'))
# Use the loaded model to predict the output for a new input.
# Here, 30 represents 30 years of Experience.
# Double square brackets are used because the model expects 2D input data.
result=loaded_model.predict([[12]])
# Display the predicted result
result
```
